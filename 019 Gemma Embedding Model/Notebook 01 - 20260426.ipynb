# **Embedding**

```python
from openai import OpenAI

# 1. Initialize the client to point to your local LM Studio server
client = OpenAI(base_url="http://localhost:1234/v1", api_key="lm-studio")

def get_embedding(text, model="local-model"):
    # Clean the text (remove newlines)
    text = text.replace("\n", " ")
    
    # 2. Request the embedding
    # The 'model' parameter can usually be anything since LM Studio 
    # uses whichever model is currently loaded in the GUI.
    response = client.embeddings.create(input=[text], model=model)
    
    return response.data[0].embedding
```

```python
# 3. Test it out
my_text = "Gemma is a lightweight, state-of-the-art open model from Google."
vector = get_embedding(my_text)

print(f"Vector Length: {len(vector)}")
print(f"First 5 dimensions: {vector[:5]}")
```
