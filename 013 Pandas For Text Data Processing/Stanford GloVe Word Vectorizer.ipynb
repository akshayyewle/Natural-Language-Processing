```python
import numpy as np

# Load pre-trained GloVe vectors (glove.6B.300d.txt should be downloaded)
glove_file = 'glove.6B.300d.txt'  # Use the path to your GloVe file
embedding_dict = {}

with open(glove_file, 'r', encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.asarray(values[1:], dtype='float32')
        embedding_dict[word] = vector
```

```python
# Convert a word to a vector
word = "king"
vector = embedding_dict.get(word)
if vector is not None:
    print(vector)
else:
    print(f"{word} not found in the GloVe vocabulary.")
```
